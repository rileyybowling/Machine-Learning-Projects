# Exercise 10:  model selection

Run the following experiments on the QSAR dataset:

* Compute the accuracy of a nonlinear SVM with Gaussian kernel with default parameters using cross validation.

* Compare the accuracy using default parameters to the accuracy obtained by performing model selection of the values of gamma and C.  I.e. run nested cross validation with grid search performed over the values of gamma and C.

* The parameter gamma is the one that affects the performance of an SVM with a Gaussian kernel the most.  With that in mind, do you get similar performance by performing grid search only on the value of gamma?
   
* What parameters are chosen when you train a GridSearchCV object on this dataset?  


```python
from sklearn.model_selection import train_test_split, cross_val_score, StratifiedKFold, GridSearchCV
from sklearn.svm import SVC
   
# data
filename='data/biodeg.csv'
data = np.loadtxt('data/biodeg.csv', delimiter = ';', dtype = 'S')
X = data[:, :-1].astype(float)
y = data[:, -1]
y = np.where(y == b"RB", 1, -1)


# default svm
svm = SVC(kernel='rbf')
cv = StratifiedKFold(n_splits=5, random_state=0, shuffle=True)
svm_default_acc = cross_val_score(svm, X, y, cv = cv, scoring = 'accuracy')
print(f"SVM accuracy with default values: {np.mean(svm_default_acc)}\n")

# grid search over C and gamma
param_grid = [
  {'C': [1, 10, 100, 1000], 
   'gamma': [0.001, 0.01, 0.1], 
   'kernel': ['rbf']}
 ]

svm_gc = GridSearchCV(SVC(), param_grid)
svm_gc.fit(X, y);

print(f"SVM accuracy with gridsearch over C and Gamma: {svm_gc.best_score_}")
print(f"The best C and Gamma values are as follows: {svm_gc.best_estimator_}\n")

# grid search over gamma
param_grid = [
  {'gamma': [0.001, 0.01, 0.1], 
   'kernel': ['rbf']}
 ]

svm_g = GridSearchCV(SVC(), param_grid)
svm_g.fit(X, y);

print(f"SVM accuracy with gridsearch over Gamma: {svm_g.best_score_}")
print(f"The best C and Gamma values are as follows: {svm_g.best_estimator_}")
```

**When running a gridsearch on only gamma values, the validation is close and comparable to that of the gridsearch on both C and gamma values. When looking for the best parameters, cross validation led to the results of C = 1000 and gamma = .0001 to get the best accuracy.**
